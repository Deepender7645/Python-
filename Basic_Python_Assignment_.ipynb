Question:- Grade Calculator      
4. Suppose a student has marks in 5 subjects. Store them in variables and calsulator the average. Based on the average, print the grade.
- 90 and above--> Grade A
- 75-89 --> Grade B
- 50-74 --> Grade c
Below 50 --> Fail

```python
# User enter the markes of 5 sub:
sub1 = int(input("Enter the marks of sub1:"))
sub2 = int(input("Enter the marks of sub2:"))
sub3 = int(input("Enter the marks of sub3:"))
sub4 = int(input("Enter the marks of sub4:"))
sub5 = int(input("Enter the marks of sub5:"))
average = (sub1+sub2+sub3+sub4+sub5)/5 # average variable write a 5 sub average:
if average >= 90:# check condition  the average grater then equal mark 90 :
  print("Grade A")
elif average >= 75:# check the condition the average grater then equal mark 75:
  print("Grade B")
elif average >= 50:# check the condition the average grater then equal mark 50:
  print("Grade C")
else: # Otherwise all the condition are false then run this condition:
  print("Fail")

```
